## **632. Book–Course Inheritance (6 Marks)**

Implement the following hierarchy:

* The **Book** class has the following data members:
  `name`, `n` (number of authors), `authors` (list of authors), `publisher`, `ISBN`, and `year`.
* The **derived class** has `course` as its data member.
* The derived class method should **override (extend)** the methods of the base class.

````python
class Book():
    def __init__(self,name,n,author,publisher,ISBN,year):
        self.name=name
        self.n=n
        self.author=author
        self.publisher=publisher
        self.ISBN=ISBN
        self.year=year
    def display(self):
        print(f"Name : {self.name}")
        print(f"Number of authors : {self.n}")
        print(f"Author : {self.author}")
        print(f"Publisher : {self.publisher}")
        print(f"ISBN : {self.ISBN}")
        print(f"Year : {self.year}")
class Page(Book):
    def __init__(self,name,n,author,publisher,ISBN,year,course):
        super().__init__(name,n,author,publisher,ISBN,year)
        self.course=course
    def display(self):
        super().display()
        print(f"Course : {self.course}")
````

````python
from abc import ABC, abstractmethod
class Employee(ABC):
  @abstractmethod
  def receive_call(self):
    pass
  
  @abstractmethod
  def end_call(self):
    pass
  @abstractmethod
  def is_free(self):
    pass
  
  @abstractmethod
  def get_rank(self):
    pass
````

````python
class Respondent:
    def __init__(self,id,name):
        self.id=id
        self.name=name
        self.rank=3
        self.free=True
    def receive_call(self):
        print(f"Call received by {self.name}")
        self.free=False
    def end_call(self):
        print("Call Ended")
        self.free=True
    def is_free(self):
        return self.free
    def get_rank(self):
        return self.rank
````

````python
class Manager:
    def __init__(self,id,name):
        self.id=id
        self.name=name
        self.rank=2
        self.free=True
    def receive_call(self):
        print(f"Call received by {self.name}")
        self.free=False
    def end_call(self):
        print("Call Ended")
        self.free=True
    def is_free(self):
        return self.free
    def get_rank(self):
        return self.rank
````

````python
class Director:
    def __init__(self,id,name):
        self.id=id
        self.name=name
        self.rank=1
        self.free=True
    def receive_call(self):
        print(f"Call received by {self.name}")
        self.free=False
    def end_call(self):
        print("Call Ended")
        self.free=True
    def is_free(self):
        return self.free
    def get_rank(self):
        return self.rank
````

````python
class Call:
    def __init__(self,id,name):
        self.id=id
        self.name=name
        self.assigned=False
````

````python
class CallHandler:
    respondents=[]
    managers=[]
    directors=[]
    def add_employee(self,employee):
        if employee.get_rank() == 3:
            self.respondents.append(employee)
        elif employee.get_rank() == 2:
            self.managers.append(employee)
        elif employee.get_rank() == 1:
            self.directors.append(employee)
    def dispatch_call(self,call):
        for emp in self.respondents + self.managers + self.directors:
            if emp.is_free():
                emp.receive_call()
                call.assigned = True
                return
        print("Sorry! All employees are currently busy.")
````

Write a program to create a **Bus** child class inheriting from **Vehicle**.

````python
class Vehicle:
    def __init__(self,vname,mileage,sc):
        self.vname=vname
        self.mileage=mileage
        self.sc=sc
    def fare(self):
        return self.sc*100
class Bus(Vehicle):
    def fare(self):
        base_fare = super().fare()
        maintenance_charge = base_fare * 0.10
        return base_fare + maintenance_charge
class Car(Vehicle):
    pass
````

## **653. Abstract Shape Class (9 Marks)**

* Create abstract class **Shape**
* Create classes **Rectangle** and **Circle**
* Implement `calculate_area()`
* Overload `>` operator to compare rectangle areas
* Print **MRO of Circle**

````python
from abc import ABC,abstractmethod
import math
````

````python
class Shape(ABC):
    @abstractmethod
    def calculate_area(self):
        pass
````

````python
class Rectangle(Shape):
    def __init__(self,l,b):
        self.length=l
        self.breadth=b
    def calculate_area(self):
        return self.length*self.breadth
    def __gt__(self,other):
        return self.calculate_area()>other.calculate_area()
class Circle(Shape):
    def __init__(self,r):
        self.radius=r
    def calculate_area(self):
        return math.pi*(self.radius**2)
    def __gt__(self,other):
        return self.calculate_area()>other.calculate_area()
````

````python
r1=Rectangle(2,3)
r2=Rectangle(4,5)
````

````python
print("Area of Rectangle 1:", r1.calculate_area())
print("Area of Rectangle 2:", r2.calculate_area())

# Compare rectangles using > operator
if r1 > r2:
    print("Rectangle 1 has greater area than Rectangle 2")
else:
    print("Rectangle 2 has greater area than Rectangle 1")
````

````python
c1 = Circle(7)
print("Area of Circle:", c1.calculate_area())
````

````python
print("MRO of Circle class:")
print(Circle.__mro__)
````

## **657. Net Salary Using Inheritance (9 Marks)**

Create classes:

* **Employee** (abstract)
* **Perks**
* **NetSalary**

Rules:

* DA = 35%
* HRA = 17%
* PF = 12%

Print employee details, perks, and final salary.

##### 
```
Write a Python Program to Find the Net Salary of Employee using Inheritance.
Create three Class Employee, Perks, NetSalary. Make an Employee class as an abstract class.
Employee class should have methods for following tasks.- To get employee details like employee id, name and salary from user. - To print the Employee details. - return Salary.- An abstract method emp_id.
Perks class should have methods for following tasks.- To calculate DA, HRA, PF. - To print the individual and total of Perks (DA+HRA-PF).
Netsalary class should have methods for following tasks.- Calculate the total Salary after Perks.- Print employee detail also prints DA, HRA, PF and net salary.
Note 1: DA-35%, HRA-17%, PF-12%
Note 2: It is compulsory to create objects and demonstrating the methods with 
              Correct output.                                                                                                                                       Example:
Employee ID: 1
Employee Name: John
Employee Basic Salary: 25000
DA: 8750.0
HRA: 4250.0
PF: 3000.0
Total S
```alary: 35000.0

````python
from abc import ABC,abstractmethod
````

````python
class Employee(ABC):
    def __init__(self):
        self.employee_id=int(input("Enter employee ID : "))
        self.name=input("Enter employee name: ")
        self.salary=int(input("Enter employee salary : "))
    def putdata(self):
        print("Employee ID:", self.employee_id)
        print("Employee Name:", self.name)
        print("Employee Basic Salary:", self.salary)

    def return_salary(self):
        return self.salary

    @abstractmethod
    def emp_id(self):
        pass
class Perks(Employee):
    def calculate(self):
        self.DA=self.salary*0.35
        self.HRA=self.salary*0.17
        self.PF=self.salary*0.12       
        self.perks=self.DA+self.HRA-self.PF
    def display_perks(self):
        print(f"DA : {self.DA}")
        print(f"HRA : {self.HRA}")
        print(f"PF : {self.PF}")
        print(f"Perks : {self.perks}")
class NetSalary(Perks):
    def emp_id(self):
        return self.employee_id
    def calculate_salary(self):
        self.calculate()
        self.net_salary = self.salary + self.perks
    def display(self):
        self.putdata()
        self.display_perks()
        print("Total Salary:", self.net_salary)
````
