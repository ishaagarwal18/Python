### Write a pager program. Your solution should prompt for a filename, and display the text file 25 times, pausing each time to ask user to end "continue" in order to show next 25 lines or enter word"stop" to close file.


````python
def pager(f):
    with open(f,encoding='utf-8') as file:
        lines=file.readlines()
        total=len(lines)
        start=0
        end=min(25,total)
        while start<total:
            for i in range(start,end):
                print(lines[i],end="")
            user=input("Enter continue or stop : ")
            if user.lower=="stop":
                break
            start=25
            end=min(start+25,total)
````

```
Write a Python program to reverse the content of a one file and store it in second file and also convert content of second 
file into uppercase and store it in third file and also count number of Vowels in third file and also print only 2nd line from 
the content of third file.
Examples:
If data file one contains the following data:
Friends are crazy, Friends are naughty !
Friends are honest, Friends are best !
Output 1:
! tseb era sdneirF ,tsenoh era sdneirF
! ythguan era sdneirF ,yzarc era sdneirF
Output 2:
! TSEB ERA SDNEIRF ,TSENOH ERA SDNEIRF
! YTHGUAN ERA SDNEIRF ,YZARC ERA SDNEIRF
Output 3:
Vowels = 22
Output 4:
! YTHGUAN ERA SDNEIRF ,YZARC ERA SDNEIRF
```

````python
%%writefile friend.txt
Friends are crazy, Friends are naughty !
Friends are honest, Friends are best !
````

````python
file=open("friend.txt")
content=file.readlines()
file1=open("reverse.txt","w")
file2=open("upper.txt","w")
file3=open("vowel.txt","w")
file4=open("second.txt","w")
s=""
vowels=0
for i in content:
    for j in i:
        if j in"aeiouAEIOU":
            vowels+=1
        s=j+s
u=s.upper()
l=u.split("\n")
print(l)
p="Vowels = "+str(vowels)
file1.write(s)
file2.writelines(u)
file3.write(p)
file4.writelines(l[2])
file1.close()
file2.close()
file3.close()
file4.close()
````

```
Write a python program to accept string/sentence from user till the user enters “END”. Each string/sentence entered by 
user should be a newline in file. Save all the lines in file and display only those lines which begin with capital letter.
Example:
Enter Something (for quit enter END):Hi Friends
Enter Something (for quit enter END):how are you all
Enter Something (for quit enter END):I am fine
Enter Something (for quit enter END):hope you all are fine
Enter Something (for quit enter END):END
The Line started with Capital Letters:

```

````python
with open("End.txt","w+") as f:
    while True:
        s=input("Enter Something (for quit enter END) : ")
        if s=="END":
            break
        f.writelines(s+"\n")
    f.seek(0)
    for i in f:
        if i[0].isupper():
            print(i)
````
