#### Q291 Write a Python program to check if a string is palindrome or not.

````python
s=input("Enter a string : ")
new=s[::-1]
if s==new:
    print(s," is palindrome")
else:
    print(s," is not a palindrome")
print(new)
````

#### The Syracuse (also called Collatz or Hailstone) sequence is generated by starting with a natural number and repeatedly 
applying the following function until reaching 1:
 ```syr(x) = 1/2 if x is even; and 
syr(x) = 3x + 1 if x is odd
For example, the Syracuse sequence starting with 5 is: 5, 16, 8, 4, 2, 1 
Use of inbuilt function or math library is not allow.
 This sequence will always go to 1 for every possible starting vae.
 Write a program at 
 1.Gets a star ng value from theser 
 2.Prints the Syracuse seque the starting `ng``` value.

````python
def syr(x):
    if x%2==0:
        return x//2
    else:
        return 3*x+1
n=int(input("Enter number : "))
while n>1:
    print(n, end=" ")
    n=syr(n)
print(n)
````

````python
# ENCRYPTION

msg = input("Enter the message: ")
key = int(input("Enter the key: "))

encrypted = ""

# group 0 takes indices: 0, key, 2*key, 3*key, ...
# group 1 takes indices: 1, key+1, 2*key+1 ...
# ... up to group (key-1)
for start in range(key):
    i = start
    while i < len(msg):
        encrypted += msg[i]
        i += key

print("Encrypted Message:", encrypted)
````
