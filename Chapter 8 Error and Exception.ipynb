 Class Order containing

````python
class Invalidtoppings(Exception):
    pass
class Invalidcheese(Exception):
    pass
class Pizza:
    def __init__(self,size,toppings,cheese):
        self.size=size
        self.cheese=cheese
        self.toppings=toppings
        self.validtoppings()
        self.validcheese()
        
    def validtoppings(self):
        validtop=['corn', 'tomato', 'onion', 'capsicum', 'mushroom', 'olives', 'broccoli']
        for topping in self.toppings:
            if topping not in validtop:
                raise Invalidtoppings(f"Invalid toppings -{topping}")
                
    def validcheese(self):
        validch=['mozzarella', 'feta', 'cheddar']
        for ch in self.cheese:
            if ch not in validch:
                raise Invalidcheese(f"Invalid cheese -{ch}")
                
    def price(self):
        sizeprice={'small':50, 'medium':100,'large':200}
        toppingprice=20
        exotic_toppingprice=50
        cheeseprice=50
        totalprice=sizeprice[self.size]
        for topping in self.toppings:
            if topping in ['mushroom', 'olives', 'broccoli']:
                totalprice+=exotic_toppingprice
            else:
                totalprice+=toppingprice
        totalprice+=len(self.cheese)*cheeseprice
        return totalprice
            
class Order:
    def __init__(self,name,customerid):
        self.name=name
        self.customerid=customerid
        self.pizzas=[]
    def order(self,pizza):
        self.pizzas.append(pizza)
    def bill(self):
        totalcost=0
        print(f"Order details of {self.name} with ID {self.customerid}")
        for index,pizza in enumerate(self.pizzas,start=1):
            print(f"Pizza {index}")
            print(f"Size: {pizza.size}")
            print(f"Toppings:{','.join(pizza.toppings)}")
            print(f"Cheese : {','.join(pizza.cheese)}")
            pizzacost=pizza.price()
            print(f"Cost:{pizzacost} Rupees")
            totalcost+=pizzacost
        print(f"Total order cost is {totalcost} Rupees")
````

````python
class WordPlay:
    def __init__(self,word):
        self.word=word
    def words_with_length(self,length):
        w=[]
        for i in self.word:
            if len(i)==length:
                w.append(i)
        return w
    def starts_with(self,char1):
        w=[]
        for i in self.word:
            if i[0]==char1:
                w.append(i)
        return w
    def ends_with(self,char2):
        w=[]
        for i in self.word:
            if i[-1]==char2:
                w.append(i)
        return w
    def palindromes(self):
        w=[]
        for i in self.word:
            if i==i[::-1]:
                w.append(i)
        return w
    def only(self,str1):
        w=[]
        s=set(str1)
        for i in self.word:
            words=set(i)
            if s==words:
                w.append(i)
        return w
    def avoids(self,str2):
        w=[]
        s=set(str2)
        for i in self.word:
            words=set(i)
            if words.intersection(s)==set():
                w.append(i)
        return w
````

Write a simple implementation of this model which clearly shows how the different classes are composed. Write some

````python
class Artist:
    def __init__(self, name):
        self.name = name
        self.albums = []
        self.songs = []

    def add_album(self, album):
        self.albums.append(album)

    def add_song(self, song):
        self.songs.append(song)

class Song:
    def __init__(self, title, artist, album, duration):
        self.title = title
        self.artist = artist
        self.album = album
        self.duration = duration

class Album:
    def __init__(self, title, artist):
        self.title = title
        self.artist = artist
        self.tracks = []

    def add_track(self, song):
        self.tracks.append(song)


class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add_song(self, song):
        self.songs.append(song)

    def display(self):
        print("Playlist:", self.name)
        for song in self.songs:
            print(f"{song.title} - {song.artist.name}")
````

````python
# Create artists
artist1 = Artist("Arijit Singh")
artist2 = Artist("Various Artists")

# Create album
album1 = Album("Soulful Hits", artist1)
artist1.add_album(album1)

# Create songs
song1 = Song("Tum Hi Ho", artist1, album1, "4:22")
song2 = Song("Channa Mereya", artist1, album1, "4:49")

# Add songs to album
album1.add_track(song1)
album1.add_track(song2)

# Add songs to artist
artist1.add_song(song1)
artist1.add_song(song2)

# Create playlist
playlist = Playlist("My Favorites")

# Add all album songs to playlist
for song in album1.tracks:
    playlist.add_song(song)

# Display playlist
playlist.display()
````
