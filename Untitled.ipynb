````python
def compare_files(file1, file2):
    with open(file1, 'r') as f1, open(file2, 'r') as f2:
        lines1 = f1.readlines()
        lines2 = f2.readlines()

    min_lines = min(len(lines1), len(lines2))

    for i in range(min_lines):
        line1 = lines1[i]
        line2 = lines2[i]

        min_len = min(len(line1), len(line2))

        for j in range(min_len):
            if line1[j] != line2[j]:
                print(f"line number {i + 1} colNo. {j + 1}")
                return

        # If one line is longer than the other
        if len(line1) != len(line2):
            print(f"line number {i + 1} colNo. {min_len + 1}")
            return

    # If one file has extra lines
    if len(lines1) != len(lines2):
        print(f"line number {min_lines + 1} colNo. 1")
    else:
        print("Files are identical")


# Example usage
compare_files("python1.txt", "python2.txt")
````
