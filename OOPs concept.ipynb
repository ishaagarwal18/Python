```python
class Mynum:
    def __init__(self):
        print("Calling the init")
        self.val=0
    def increment(self):
        self.val+=1
        print(self.val)
```

```python
class enc:
    def __init__(self,a,b):
        self.a=a
        self.b=b
    def add(self):
        return self.a*self.b
```

```python
enc_object=enc(3,4)
```

```python
enc_object.add()
```

```python
class employee:
    empcount=0
    def __(self,name,salary):
        self.name=name
        self.salary=salary
        employee.empcount+=1
    def displaycount(self):
        print(f"Total employees are {employee.empcount}")
    def displayemployee(self):
        print(f"Name is {self.name} and salary is {self.salary}")
    def __str__(self):
        return f"The employee name is {self.name}"
```

```python
p=employee("python",10000)
```

```python
p.displaycount()
```

```python
p.displayemployee()
```

```python
print(employee.empcount)
```

```python
print(p)
```

```python
class car:
    def __init__(self,color,price):
        self.color=color
        self.price=price
    def __str__(self):
        return f"Color of car is {self.color} and price is {self.price}"
```

```python
maruti=car("Black",1000000)
kia=car("Red",1700000)
```

```python
print(maruti)
```

```python
print(kia)
```
