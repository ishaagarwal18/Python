````python
%%writefile stringmodule.py
def stringlength(s):
    return len(s)-s.count(" ")
def stringreverse(s):
    return s[::-1]
````

````python
import stringmodule
````

````python
print(stringmodule.stringlength("   Hello   "))
````

````python
print(stringmodule.stringreverse("Bye"))
````
