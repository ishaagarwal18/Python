```python
from bs4 import BeautifulSoup
with open("newflipkart.html",encoding="UTF-8") as f:
    html=f.read()
```

```python
soup=BeautifulSoup(html,"html.parser")
data=soup.find_all("div",class_="tUxRFH")
name=[]
rom=[]
color=[]
stars=[]
review=[]
old_price=[]
new_price=[]
discount=[]
exchange=[]
for i in data:
    n1=i.find("div",class_="KzDlHZ").text
    if(n1):
        n=n1.split("(")
        name.append(n[0])
        n2=n[1].split(",")
        color.append(n2[0])
    else:
        name.append(0)
        color.append(0)
    s=i.find("div",class_="XQDdHH").text
    stars.append(s)
    r=i.find("span",class_="Wphh3N").text.split("&")
    review.append(r[1].split(" ")[0])
    op=int(i.find("div",class_="yRaY8j ZYYwLA").text.lstrip("₹").replace(",",""))
    old_price.append(op)
    np=int(i.find("div",class_="Nx9bqj _4b5DiR").text.lstrip("₹").replace(",",""))
    new_price.append(np)
    dis=i.find("div",class_="UkUFwK").text.split(" ")
    discount.append(dis[0])
    e=i.find_all("div",class_="yiggsN O5Fpg8")[-2]
    for k in e:
        if(k.text=="Saver Deal"):
            exchange.append(k.replace("Saver Deal","NaN"))
        else:
            exchange.append(k.text.lstrip("₹").replace(",",""))
    ro=i.find_all("ul",class_="G4BRas")
    for j in ro:
        if j:
            r1=j.text.split(" ")
            rom.append(r1[0]+" "+r1[1])
        else:
            rom.append(0)
        
    

import pandas as pd
d={
    "Name":name,
    "Color":color,
    "Stars":stars,
    "Rom":rom,
    "Reviews":review,
    "Old price":old_price,
    "New price":new_price,
    "Discount":discount,
    "Exchange Value":exchange
    
}
df=pd.DataFrame(d)
```

```python
df["Difference"]=df["Old price"]-df["New price"]
```
