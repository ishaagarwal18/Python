##### np.concatenate
```
.It joins two given array at a given axis
```

````python
c=np.arange(6).reshape(2,3)
d=np.arange(6,12).reshape(2,3)
print(c)
print(d)
````

````python
np.concatenate((c,d))
````

````python
np.concatenate((c,d),axis=0)
````

````python
np.concatenate((c,d),axis=1)
````

##### np.sort
```
. It sorts the given array based on the given asix

````python
class Test:
    def __init__(self):
        self.x = 0
class Derived_Test(Test):
    def __init__(self):
        self.y = 1
def main():
     b = Derived_Test()
     print(b.x,b.y)
main()
````

````python
class Demo:
    def __init__(self):
        self.x = 1
    def change(self):
        self.x = 10
class Demo_derived(Demo):
    def change(self):
        self.x=self.x+1
        return self.x
def main():
    obj = Demo_derived()
    print(obj.change())

main()
````
